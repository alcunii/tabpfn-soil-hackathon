## Evidence available right now

Load every result JSON and print the headline number per cell. This is the honest,
measured state of the project.

```python
import json, glob, os
ROOT = os.path.abspath(os.path.join(os.getcwd(), '..'))
if not os.path.isdir(os.path.join(ROOT, 'results')):
    ROOT = os.getcwd()
rows = []
for f in sorted(glob.glob(os.path.join(ROOT, 'results', 'feas_e*.json'))):
    d = json.load(open(f))
    if d['task'].startswith('crop'):
        rows.append((d['cell'], d['n_context'], 'acc=%.4f' % d['accuracy'], 'ece=%.4f' % d['ece']))
    else:
        rows.append((d['cell'], d['n_context'], 'r2=%.3f' % d['r2'], 'rpd=%.2f(%s)' % (d['rpd'], d['rpd_label'])))
for r in rows:
    print(f'{r[0]:<26} ctx={r[1]:>5}  {r[2]:<12} {r[3]}')
```
